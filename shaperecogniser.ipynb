```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow




def polygon(path):
  # Load the original image
  image = cv2.imread(path)
  image = cv2.resize(image, (500,500))
  grey_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


  # Parameters for function
  maxCorners = 4
  qualityLevel = 0.01
  minDistance = 200

  # Detecting corners
  corners = cv2.goodFeaturesToTrack(grey_image, maxCorners, qualityLevel, minDistance)
  corners = np.intp(corners)
  corn_count = len(corners)

  # Draw circles around the corners
  for corner in corners:
      x, y = corner.ravel() #bring out the values of the array to get coordinates of the corners
      cv2.circle(image, (x, y), 3, (0, 0, 255), -1)

  #corner display
  cv2_imshow(image)
  cv2.waitKey(0)
  cv2.destroyAllWindows()

  print("Thee number of corners is",corn_count)

  if corn_count == 3:
    print("The image is of a triangle")
  elif corn_count == 4:
    print("The image is of a square")
  else:
    print("Rerun the code")
```
